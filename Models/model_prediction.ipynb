```python
import joblib
import pandas as pd
```

```python
df = pd.read_csv('test.csv')
ids = df['ID']            
X = df.drop(columns=['ID']).values
model = joblib.load('HNSC/models_integrated/MLP_cluster.pkl')
y_pred = model.predict(X)
y_prob = model.predict_proba(X)[:, 1]
result = pd.DataFrame({
    'ID': ids,
    'y_prob': y_prob
})
result.to_csv('prediction.csv', index=False)
```
